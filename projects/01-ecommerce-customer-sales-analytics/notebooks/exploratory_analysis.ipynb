```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load cleaned dataset
df = pd.read_csv('../data/processed/ecommerce_transactions_cleaned.csv')
df['order_date'] = pd.to_datetime(df['order_date'])
print(f'Loaded {len(df)} transactions.')
df.head()
```

```python
# Overall Executive KPIs
total_sales = df['sales'].sum()
total_profit = df['profit'].sum()
margin = (total_profit / total_sales) * 100
print(f'Total Sales: ${total_sales:,.2f}')
print(f'Total Profit: ${total_profit:,.2f}')
print(f'Profit Margin: {margin:.2f}%')
```

```python
# Category Performance Breakdown
cat_perf = df.groupby('category').agg({'sales': 'sum', 'profit': 'sum', 'order_id': 'count'})
cat_perf['margin_pct'] = (cat_perf['profit'] / cat_perf['sales']) * 100
cat_perf.round(2)
```
