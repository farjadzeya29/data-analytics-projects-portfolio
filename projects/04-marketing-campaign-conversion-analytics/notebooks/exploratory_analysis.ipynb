```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/marketing_leads_cleaned.csv')
print(f'Loaded {len(df)} leads.')
df.head()
```

```python
# Conversion Funnel Volume
print('Leads:', len(df))
print('MQL:', df['is_mql'].sum())
print('SQL:', df['is_sql'].sum())
print('Opp:', df['is_opportunity'].sum())
print('Won:', df['is_customer'].sum())
```

```python
# Channel Performance
ch_summary = df.groupby('marketing_channel').agg(leads=('lead_id', 'count'), won=('is_customer', 'sum'), revenue=('deal_value_usd', 'sum'))
ch_summary.round(2)
```
