```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/supply_chain_shipments_cleaned.csv')
print(f'Loaded {len(df)} commercial shipments.')
df.head()
```

```python
# Network OTD Benchmark
otd = df['on_time_flag'].mean() * 100
avg_delay = df[df['delivery_status'] == 'Delayed']['delivery_delay_days'].mean()
print(f'Network OTD: {otd:.2f}%')
print(f'Average Delay Duration: {avg_delay:.2f} days')
```

```python
# Warehouse Hub Fulfillment
wh_summary = df.groupby('origin_warehouse').agg(shipments=('shipment_id', 'count'), ontime=('on_time_flag', 'sum'))
wh_summary['otd_pct'] = (wh_summary['ontime'] / wh_summary['shipments']) * 100
wh_summary.round(2)
```
