```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/banking_churn_cleaned.csv')
print(f'Loaded {len(df)} customer records.')
df.head()
```

```python
# Overall Churn Rate
churn_rate = df['exited'].mean() * 100
print(f'Overall Customer Attrition: {churn_rate:.2f}%')
```

```python
# Multi-Product Paradox
prod_churn = df.groupby('num_of_products').agg(total=('customer_id', 'count'), churned=('exited', 'sum'))
prod_churn['rate'] = (prod_churn['churned'] / prod_churn['total']) * 100
prod_churn.round(2)
```

```python
# Active vs Inactive Member Attrition
df.groupby('is_active_member')['exited'].mean() * 100
```
